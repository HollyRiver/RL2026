## Markov Process

### **1. Agent & Environment**

`-` Reinforcement Learning

* Agent를 제외한 모든 것들을 Environment라고 함
* Environment는 Action이 주어지면 State와 Reward를 Agent에 부여
* Tick 또는 Time-Step마다 해당 루프가 수행됨

### **2. Unified notation**

`-` 에피소드가 종료되는 상황을 특별한 상태가 들어와서 자기 자신으로 전이하면서 0의 보상값을 발생시키는 것으로 생각

* 단일 표기법으로 표기 가능: 몰?루

### **3. Markov Process**

* 강화학습은 순차적 의사결정 문제를 푸는 방법론

$$\text{MP} == (S, P)$$

* 상태의 집합 $S$
> 가능한 상태를 모두 모아 놓은 집합
* 전이확률 행렬 $P$
> 각 상태로의 전이확률의 행렬

### **4. Markov Reward Process**

* 상태로 이동할 확률에 보상이 추가됨


$$\text{MRP} == (S, P, R, \gamma)$$

* 보상함수 $R$
> 기댓값으로 설정. 특정 상태에 도달했을 때 받는 보상이 매번 조금씩 다를 수 있음
* 할인률 Discount Rate $\gamma$
> 미래에 얻을 보상에 비해 당장 얻는 보상을 얼마나 더 중요하게 여길 지에 대한 인자

`-` 상태와 보상

* 상태가 바뀔 때마다 해당하는 보상을 얻음
* 누적 보상의 합을 최대화: 리턴 $G_t$를 최대화
> $\gamma$값에 따라 리턴이 수렴하도록 만듦

`-` 각 상태의 가치(Value) 평가

* 상태 하나를 숫자 하나로 평가: 이후 시점에 대해서 보상을 더하여 기댓값을 측정
* 에피소드 샘플링에 따라 가치 추정량이 달라지므로 Monte-Carlo 접근법으로 추정: 샘플링 사용

`-` State-Value Function

* 상태를 입력하면 그 상태의 가치를 출력하는 함수
> 에피소드마다 리턴이 다르기 때문에 어떤 상태 $s$의 가치 $v(s)$는 기댓값을 이용하여 정의 $v(s) = \mathbb E[G_t~|~S_t = s]$

> 보통 Goal의 Reward를 가장 높게 설정하는 편. Goal과의 Path가 길 때, Discount Rate가 초반 부분에 작게 적용되다보니 방향성을 짐작하지 못하는 경우가 많아짐.
>
> 복잡한 패턴이 없는 경우에는 마지막 Goal에만 Reward를 설정해도 충분함. 특별한 게 아닌 이상 나머지 상태는 똑같이 Reward 설정하거나 조금씩만 조정.
