## TD Learning

### **1. SARSA**

* state-action -> state-action으로의 전이를 고려
* 어떤 행동을 해야 하는지에 대한 문제

`-` Algorithm

* 매 step마다 가치 함수 업데이트
* 예측된 가치를 최대화하는 행동을 수행하도록 Policy 설정

`-` Optimal Policy

* Infinite Exploration을 보장해야 함
* Greedy in 뭐시기

### **2. Q-Learning**

$$Q(S_t, A_t) \leftarrow Q(S_t, A_t) + \alpha [R_{t+1} + \gamma \max_a Q(S_{t+1}, a) - Q(S_t, A_t)]$$

* SARSA는 다음 상태에서 실제 행동을 통해 가치 업데이트
* Q-Learning은 다음 상태에서 실제로 무엇을 할 것인지와 상관없이 그 상태에서 선택할 수 있는 가장 높은 가치를 가진 행동의 값을 가져와 업데이트

### **3. Games, Afterstates, and Other Special Cases**

* 상태 가치 $V(s)$: 상태 자체의 가치
* action value $Q(s, a)$: 상태에서 행동의 가치. 확률론적?
* afterstate value $V(a, s')$: 행동으로 만들어지는 직후의 상태의 가치. 결정론적?
> 환경에 대한 모든 것을 알고 있어야 함. $s'$을 확실하게 알고 있어야 함.
