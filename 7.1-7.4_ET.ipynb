## Eligibility Traces

### **1. n-step TD Prediction**

* non-terminate episodes에 적용 가능하므로, Online-Learning에 유리함
* n-step TD Prediction: n-step 이후의 reward까지 고려해서 현재 가치를 평가함

$$G_{t:t+n}=R_{t+1}+\gamma R_{t+2} + \cdots + \gamma^{n-1} R_{t+n} + \gamma^n V_{t+n-1}(S_{t+n})$$

`-` Effects

* Lookahead: t+n 시점까지 기다려야 함
* Error reduction: MC 방법보다 분산이 감소

### **2. n-step SARSA Control**

* TD 방식에서 $Q(S, A)$로 확장...
* Rolling buffer: dequeue 자료구조가 필요
* Tail processing

### **3. Purpose of n-step TD's eligibility trace**

* TD($\lambda$)

`-` $\lambda$-Return

$$G_t^{\lambda}=(1-\lambda)\sum_{n=1}^{\infty}\lambda^{n-1}G_{t:t+n}$$

`-` Effects

* non-causal: 모든 미래의 데이터 필요
* Trace-off control by $\lambda$: bias / variance 통제
* Out of Algorithm Boundary: 뭐시기

`-` Backward view of TD($\lambda$): Eligibility Traces

* 현 시점에서 과거를 보는 관점.
* 실시간으로 현재 상태에 대한 갱신이 가능...
* 과거에 얼마나 관측이 많이 되었는지를 고려하는 방식(Traces)

### **4. True-Online TD($\lambda$) for better equivalence**

* 업데이트 과정에서 correction term 추가

$$V_{t+1}(s)=V_t(s) + \alpha \delta_t E_t(s) + \alpha (V_t(S_t)-V_{t-1}(S_t)(E_t(s)-I){s S_t})$$
