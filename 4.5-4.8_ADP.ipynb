## **Asynchronous Dynamic Programming**

### **1. Asynchronous DP**

* 상태 집합 전체에 대해 체계적인 일괄 계산을 수행하지 않는 제자리 반복 DP 알고리즘
* 특정 상태를 편의적으로 업데이트 가능

$$V_{k+1}(s) = \max_{a} \sum_{s', r} p(s', r~|~s, a)[r + \gamma V_k (s')], ~~ \text{if} ~ s = s_k$$

$$V_{k+1}(s) = V_k(s), ~~ \text{if} ~ s \neq s_k$$

> 상태를 업데이트하는 순서나 빈도는 자유롭게 정할 수 있지만, 0번으로 만드는 상태가 있어서는 안됨. 최소 한 번은 업데이트를 해야 함.

### **2. Generalized Policy Iteration**

* 정책 평가(Policy Evaluation): 가치 함수를 현재 정책에 일관되게 만드는 과정. $V \to v_{\pi}$
* 정책 개선(Policy Improvement): 현재 가치 함수에 대해 현재 정책을 그리디하게 만듦. $\pi \to \text{greedy}(V)$
