## **Dynamic Programming**

### **1. DP**

* MDP의 전이확률을 모두 알고 있을 때 사용 가능


`-` Policy Evaluation

* Value Function을 업데이트

$$\begin{align}
v_{k+1}(s) & = \mathbb E_{\pi} [R_{t+1} + \gamma v_k (S_{t+1})~|~S_t = s] \\
& = \sum_a \pi(a~|~s) \sum_{s', r} p(s', r~|~s, a) [r + \gamma v_k(s')]
\end{align}$$

### **2. Policy Improvement**

* 임의의 결정론적인 policy $\pi$를 갱신
* action-value function을 고려: $q_{\pi}(s, a)$
> 상태 s에서 a를 취해보고 기존 것보다 좋다면 policy를 바꿈
>
> $s$에서 $s'$으로 가는 확률을 보고 Policy를 결정

$$q_{\pi}(s, \pi'(s)) \geq v_{\pi}(s) \Rightarrow v_{\pi'}(s) \geq v_{\pi}(s)$$

> Policy와 Transition Matrix가 있어야 함
