## Markov Decision Process

### **1. Finite MDP**

* MDP: $p(s', r~|~s, a) = \text{Pr}[R_{t+1}=r, S_{t+1} = s'~|~S_t = s, A_t = a]$
> 현재 보상은 이전 상태와 행동에 영향을 받음
* 일반적인 MP는 아님

### **2. Dynamics of Environment**

* MDP $p$를 알고 있다면 기대 보상 등등 많은 것들을 알 수 있음 -> 상태의 가치 평가 등

* $r(s, a) = \sum_{r \in R} \sum_{s' \in S} r\cdot p(s', r~|~s, a)$
> s에서 a를 취했을 때 받을 수 있는 기대 보상
>
> 환경이 확률적으로 작동하더라도 그 확률 분포를 알고 있다면 기대 보상을 예측할 수 있음

### **3. Derived Quantities**

`-` 상태 전이 확률

* 보상 $R_{t+1}$에 상관 없는 다음 상태 $S_{t+1}$의 확률: $p(s'~|~s, a) = \sum_{r \in R}p(s', r ~|~s, a)$
> 에이전트가 환경 내에서 어떻게 움직이는지를 수학적으로 기술하는 수식
>
> 보상에 대한 정보는 필요 없고 오직 상태의 변화만 관심이 있을 때 사용

`-` 상태-행동-다음 상태에 대한 기대보상

* $p(s', r~|~s, a)$
* $r(s, a, s') = \frac{\sum_{r \in R} r \cdot p(s', r ~|~s, a)}{p(s'~|~s, a)}$
> $r(s, a)$와 달리 다음 상태까지 결정되었을 때의 확률임. 다음 상태가 확률적으로 변할 수 있음...
