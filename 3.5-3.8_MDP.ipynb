## Markov Decision Process

### **1. Finite MDP**

* MDP: $p(s', r~|~s, a) = \text{Pr}[R_{t+1}=r, S_{t+1} = s'~|~S_t = s, A_t = a]$
> 현재 보상은 이전 상태와 행동에 영향을 받음
* 일반적인 MP는 아님

### **2. Dynamics of Environment**

* MDP $p$를 알고 있다면 기대 보상 등등 많은 것들을 알 수 있음 -> 상태의 가치 평가 등

* $r(s, a) = \sum_{r \in R} \sum_{s' \in S} r\cdot p(s', r~|~s, a)$
> s에서 a를 취했을 때 받을 수 있는 기대 보상
>
> 환경이 확률적으로 작동하더라도 그 확률 분포를 알고 있다면 기대 보상을 예측할 수 있음

### **3. Derived Quantities**

`-` 상태 전이 확률

* 보상 $R_{t+1}$에 상관 없는 다음 상태 $S_{t+1}$의 확률: $p(s'~|~s, a) = \sum_{r \in R}p(s', r ~|~s, a)$
> 에이전트가 환경 내에서 어떻게 움직이는지를 수학적으로 기술하는 수식
>
> 보상에 대한 정보는 필요 없고 오직 상태의 변화만 관심이 있을 때 사용

`-` 상태-행동-다음 상태에 대한 기대보상

* $p(s', r~|~s, a)$
* $r(s, a, s') = \frac{\sum_{r \in R} r \cdot p(s', r ~|~s, a)}{p(s'~|~s, a)}$
> $r(s, a)$와 달리 다음 상태까지 결정되었을 때의 확률임. 다음 상태가 확률적으로 변할 수 있음...

### **4. Value Function**

`-` State-Value Function

$$v_{\pi}(s) = \mathbb E_{\pi} [G_t ~|~ S_t = s] = \mathbb E_{\pi}[\sum_{k=0}^{\infty} \gamma^{k}R_{t+k+1}~|~ S_t = s]$$

* $R_{t+k+1}$: 시점 $t$로부터 $k+1$단계 후에 수신하는 보상값
* action과 state를 모두 고려한 값인듯?

`-` Action-Value Function

$$q_{\pi}(s, a) = \mathbb E_{\pi} [G_t ~|~ S_t = s, A_t] = \mathbb E_{\pi} [\sum_{k=0}^{\infty} \gamma^k R_{t+k+1} ~|~ S_t = s, A_t = a]$$

* 상태 $s$에서 행동 $a$를 취했을 때의 가치

### **5. Bellman Equation**

$$\begin{align}
v_{\pi}(s) & = \mathbb E_{\pi} [G_t ~|~ S_t = s] \\
& = \mathbb E_{\pi} \left[ \sum_{k=0}^{\infty} \gamma^k R_{t+k+1} ~|~ S_t = s \right] \\
& = \sum_{a} \pi(a~|~s) \sum_{s'}\sum_r p(s', r~|~s, a) \left[ r + \gamma \mathbb E_{\pi} \left[ \sum_{k=0}^{\infty} \gamma^k R_{t+k+1} ~|~ S_{t+1} = s' \right] \right] \\
& = \sum_{a} \pi(a~|~s) \sum_{s', r} p(s', r~|~s, a) \left[ r + \gamma v_{\pi}(s') \right]
\end{align}$$

* 현 상태에서 policy $\pi$에 의해 모든 가능한 행동과 상태, 보상을 고려하여 기댓값을 취함
* 발생 가능한 모든 궤적에 대해 평균을 구하는 식. 각 궤적이 발생할 확률을 가중치로 하여 합산

### **6. Optimal Value Function**

`-` 모든 Policy 중에 Optimal한 것

* $v_*(s) = \max_{\pi} v_{\pi}(s)$
* $q_*(s, a) = \max_{\pi} q_{\pi} (s, a)$

$$q_*(s, a) = \mathbb E[R_{t+1} + \gamma v_* (S_{t+1}~|~S_t = s, A_t = a)]$$

`-` Bellman Optimality Equation

$$\begin{align}
v_*(s) & = \max_{a \in \mathcal A(s)} q_{\pi_*}(s, a) \\
& = \max_a \mathbb E [R_{t+1} + \gamma v_* (S_{t+1}~|~S_t = s, A_t = a)] \\
& = \max_{a \in \mathcal A (s)} \sum_{s', r} p (s', r~|~s, a) [r + \gamma v_*(s')]
\end{align}$$
