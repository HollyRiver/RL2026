## Monte Carlo

### **1. Monte Carlo Prediction**

* 강화학습에서 몬테-카를로 방식을 적용시키면 전체 환경을 볼 수 없음
> 환경에서 시스템을 실행시키는 범위만 시뮬레이션. 시퀀스를 샘플링

`-` Model-free RL

* Dynamics에 대한 정보가 없음: Transition Probablity를 알지 못함
* 경험으로부터 학습: 환경과 상호작용함으로써 학습

`-` MC Prediction Method

* 모든 에피소드에서 얻은 리턴값을 평균

`-` First-visit Prediction

* 한 개의 에피소드에서 각 상태에 첫 번째 방문했을 때만 추정에 반영함
* Every-visit Prediction 방식도 사용됨

`-` Effect of MC Prediction

* Batch Update: 에피소드가 끝나야 업데이트 수행. Transaction unit 단위로 처리
* No Bootstrap: 관측된 리턴값을 바로 사용. 순수 Monte-Carlo Sampling에서는 부스트랩을 사용하지 않음

`-` Model-free Monte-Carlo Estimation's Traits

* Model-based RL: 환경이 어떻게 변화할 것인지에 대한 수식을 가지고 에피소드 진행
* Model-free RL: 환경 변화 없이 어쩌구...
> Greedy하게 행동만 하면, 방문하지 못한 상태와 행동이 존재하여 최적해를 찾지 못할 가능성이 존재 -> 시작 상태의 다양성을 주고, Stochastic policy를 주어 해결할 수 있음

### **2. Monte Carlo Control**

* Control: Value Estimation 이후 Policy를 개선하는 과정
* GPI Generalized Policy Iteration Process
> Evaluation-Improvement 과정을 반복하여 Policy 개선. 현재 Policy에서 Q(s, a)를 학습하고, 그리디하게 Policy 업데이트

`-` $\epsilon$-Greedy Policy

* $1-\epsilon$ 확률로 exploitation
* $\epsilon$ 확률로 exploration
